# CK1α–lenalidomide–CRBN: Wild type vs V388I mutant (3 runs)
Each replicate (run 1, 2, 3) is analysed independently: every run's wild type is
compared to its own mutant. Same graph set repeated per run. All logic lives in
`analysis_functions.py`; edit only Cell 2.


```python
import os
from analysis_functions import *
from IPython.display import display          # so tables show inside loops
apply_style()

# ===== SHARED SETTINGS =====
STRIDE         = 10          # 10 for testing; 1 for final numbers
TOTAL_TIME_NS  = 50.0
LIGAND_RESNAME = "MOL"
CHAIN_A_FRAG   = 0
HBOND_MOTIF    = ("ILE", "THR", "ASN")
SITE_CUTOFF    = 10.0
POCKET_CUTOFF  = 10.0

RUNS    = [1, 2, 3]
BASE    = "/home/lingyun/ternary_complex_md"
WT_SYS  = "CK1a_len_CRBN"           # wild type
MUT_SYS = "CK1a_len_CRBNV388I"      # mutant

def run_paths(sys_name, n):
    d = os.path.join(BASE, sys_name, f"run{n}", "prep", "equilibration", "04_production")
    return os.path.join(d, "output.parm7"), os.path.join(d, "output.dcd")

# ----- display -----
NAME_WT, NAME_MUT = "Wild type", "V388I mutant"
COL_LABELS   = {"complex": "Complex", "chain_A": "CK1a", "chain_B": "CRBN"}
CHAIN_LABELS = {"chain_A": "CK1a", "chain_B": "CRBN"}
RMSD_COLUMNS = ["complex", "chain_A", "chain_B"]
# ===========================
```

## Process all runs
Six independent universes: WT and mutant for run 1, 2, 3. Stored in `wt[n]` / `mut[n]`.


```python
wt, mut = {}, {}
for n in RUNS:
    wt_top,  wt_traj  = run_paths(WT_SYS,  n)
    mut_top, mut_traj = run_paths(MUT_SYS, n)
    wt[n]  = analyse_system(wt_top,  wt_traj,  STRIDE, TOTAL_TIME_NS,
                            f"analysis_{WT_SYS}_run{n}",  f"{WT_SYS}_run{n}",
                            LIGAND_RESNAME, CHAIN_A_FRAG)
    mut[n] = analyse_system(mut_top, mut_traj, STRIDE, TOTAL_TIME_NS,
                            f"analysis_{MUT_SYS}_run{n}", f"{MUT_SYS}_run{n}",
                            LIGAND_RESNAME, CHAIN_A_FRAG)
```

## RMSD 1 — internal fold stability (each group fit on itself)


```python
for n in RUNS:
    plot_rmsd_compare(wt[n]["rmsd_self"], wt[n]["times_ns"],
                      mut[n]["rmsd_self"], mut[n]["times_ns"],
                      RMSD_COLUMNS, COL_LABELS, NAME_WT, NAME_MUT,
                      suptitle=f"Run {n} — RMSD (each group fit on itself)")
    display(compare_means(wt[n]["rmsd_self"], mut[n]["rmsd_self"],
                          RMSD_COLUMNS, COL_LABELS, NAME_WT, NAME_MUT))
```

## RMSD 2 — drift from the start (common whole-protein alignment)


```python
for n in RUNS:
    plot_rmsd_compare(wt[n]["rmsd_frame0"], wt[n]["times_ns"],
                      mut[n]["rmsd_frame0"], mut[n]["times_ns"],
                      RMSD_COLUMNS, COL_LABELS, NAME_WT, NAME_MUT,
                      suptitle=f"Run {n} — RMSD vs frame 0 (common alignment)")
    display(compare_means(wt[n]["rmsd_frame0"], mut[n]["rmsd_frame0"],
                          RMSD_COLUMNS, COL_LABELS, NAME_WT, NAME_MUT))
```

## RMSF — per-residue flexibility


```python
for n in RUNS:
    plot_rmsf_compare(wt[n]["rmsf"], mut[n]["rmsf"], CHAIN_LABELS,
                      NAME_WT, NAME_MUT, suptitle=f"Run {n} — Per-residue RMSF")
```

## Interface hydrogen bonds — CK1α Ile-Thr-Asn → CRBN


```python
for n in RUNS:
    wt_trip = find_motif_triplet(wt[n]["u"], wt[n]["sels"]["A"], wt[n]["sels"]["B"], HBOND_MOTIF)
    s_wt, series_wt, lab_wt = hbonds_for_residues(
        wt[n]["u"], wt[n]["sels"]["A"], wt[n]["sels"]["B"], wt_trip)
    plot_hbonds(series_wt, wt[n]["times_ns"], lab_wt, title=f"Run {n} — Interface H-bonds: {NAME_WT}")

    mut_trip = find_motif_triplet(mut[n]["u"], mut[n]["sels"]["A"], mut[n]["sels"]["B"], HBOND_MOTIF)
    s_mut, series_mut, lab_mut = hbonds_for_residues(
        mut[n]["u"], mut[n]["sels"]["A"], mut[n]["sels"]["B"], mut_trip)
    plot_hbonds(series_mut, mut[n]["times_ns"], lab_mut, title=f"Run {n} — Interface H-bonds: {NAME_MUT}")

    display(s_wt[["bond","occupancy","mean_len"]].merge(
        s_mut[["bond","occupancy","mean_len"]], on="bond", how="outer",
        suffixes=("_wt","_mut")).round(2))
```

## Binding-site RMSD (site defined on each run's WT frame 0, applied to its mutant)


```python
for n in RUNS:
    site = define_binding_site(wt[n]["u"], wt[n]["sels"], SITE_CUTOFF, frame=0)
    for resids, title in [(site["CK1a"], "CK1a binding site"),
                          (site["CRBN"], "CRBN binding site"),
                          (site["CK1a"] + site["CRBN"], "Entire binding site")]:
        plot_site_rmsd(rmsd_of_resids(wt[n]["u"], resids), wt[n]["times_ns"],
                       rmsd_of_resids(mut[n]["u"], resids), mut[n]["times_ns"],
                       f"Run {n} — {title} ({SITE_CUTOFF:.0f} Å, defined on WT)")
```

## Key literature contacts (Pro352, Gly40, 388↔Gly40 clash, pocket)


```python
for n in RUNS:
    contacts = literature_contact_distances(wt[n]["u"], wt[n]["sels"])
    summary = plot_contacts_combined(contacts, wt[n]["u"], wt[n]["times_ns"],
                                     mut[n]["u"], mut[n]["times_ns"], NAME_WT, NAME_MUT)
    print(f"Run {n} contacts:")
    display(summary)
```
